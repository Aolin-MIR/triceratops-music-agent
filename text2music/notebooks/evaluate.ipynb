```python
import music21

xml_path = "/gpfs/scratch/acw769/text2score/text2score_ablated_v2/57/xml/57_output.xml"
expected_key = 'Ab major'
score = music21.converter.parse(xml_path)

# # 1. TIME SIGNATURE VALIDATION
# all_ts = score.flatten().getElementsByClass(music21.meter.TimeSignature)
# ts_actual = all_ts[0] if len(all_ts) > 0 else None
# ts_score = 1 if (ts_actual and ts_actual.ratioString == expected_ts) else 0

# 2. KEY SIGNATURE VALIDATION (Enhanced for Relative Keys)
all_ks = score.flatten().getElementsByClass(music21.key.KeySignature)
ks_actual = all_ks[0] if len(all_ks) > 0 else None

key_score = 0
if ks_actual:
    # Convert the signature to a Key object (usually defaults to Major)
    actual_key_obj = ks_actual.asKey()
    
    # Get the primary detected name (e.g., "G major")
    actual_name = actual_key_obj.name.lower()
    
    # Get the relative key name (e.g., "E minor")
    relative_key_obj = actual_key_obj.relative
    relative_name = relative_key_obj.name.lower()
    
    # Cross-validate: Match if metadata matches either the detected or relative key
    if actual_name == expected_key.lower() or relative_name == expected_key.lower():
        key_score = 1
```
